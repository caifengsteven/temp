```python
from matplotlib import pyplot as plt
import numpy as np
import os
from model import *
```

```python
#Load patterns
pattern_dir = 'train_data/Pattern'
pattern_list = []
pattern_files = sorted(os.listdir(pattern_dir), key=lambda x : int(x.split('.')[0]))
for file in pattern_files:
    with open(os.path.join(pattern_dir,file)) as f:
        pattern_list.append(np.array([float(x) for x in f.read().strip().split('\n')], dtype=np.float32))
patterns = np.vstack(pattern_list)
```

```python
# Draw patterns
col = 4
row = int((patterns.shape[0] + col - 1) / col)
for i in range(patterns.shape[0]):
    plt.subplot(row,col,i+1)
    plt.plot(patterns[i])
    plt.title(str(i+1))
plt.tight_layout()
```

```python
X = np.concatenate((patterns, 1 - patterns), axis=1)
Y = np.array(range(X.shape[0])).reshape((12,))
classify_net = ClassifyNetowrk()
```

```python
# Draw train set
col = 4
row = int((X.shape[0] + col - 1) / col)
for i in range(X.shape[0]):
    plt.subplot(row,col,i+1)
    plt.plot(X[i])
    plt.title(str(i+1))
plt.tight_layout()
```

```python
# train classify net
losses = classify_net.train(X, Y, verbose=True)
Y_hat = classify_net.predict(X)
```

```python
# draw loss
plt.plot(losses)
plt.title('Loss during training')
# show predicted result as one-hot vector, the index of which value close to 1 is the calssified pattern id
for i in range(12):
    for j in range(12):
        print('{:.4f}'.format(Y_hat[i, j]), end='\t')
    print('\n')
print(np.eye(12))
print('predict\t', np.argmax(Y_hat, axis=1))
print('truth\t', Y)
```

```python
predict_net = PredictNetwork()
```

```python
status = np.eye(patterns.shape[0])
with open('train_data/trend.txt') as f:
    trend = [int(x.strip()) for x in f.readlines()]
trend = np.array(trend).reshape((12,))
```

```python
# train predict net
losses = predict_net.train(status, trend, verbose=True)
predict_trend = predict_net.predict(status)
```

```python
# show losses during training
plt.plot(losses)
plt.title('Loss during training')
# show predict trend and compare with ground truth
print('As you can see, the predict result is consistent with the truth')
print('predict result\n', predict_trend)
print('predict trend\n', np.argmax(predict_trend, axis=1))
print('ground truth\n', trend)
```

```python
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.ticker import LinearLocator, FormatStrFormatter

def display_weights_in_3d(weights, title):
    X, Y = np.arange(weights.shape[1]), np.arange(weights.shape[0])
    X, Y = np.meshgrid(X, Y)
    fig = plt.figure()
    ax = Axes3D(fig)
    surf = ax.plot_surface(X, Y, weights, rstride=1, cstride=1, cmap=cm.viridis)
    fig.colorbar(surf, shrink=0.5, aspect=5)
    plt.title(title)
    plt.show()
```

```python
display_weights_in_3d(classify_net_weights[0].T, 'classify_net_hidden_layer')
display_weights_in_3d(classify_net_weights[1].T, 'classify_net_output_layer')
display_weights_in_3d(predict_net_weights[0].T, 'predict_net_hidden_layer')
display_weights_in_3d(predict_net_weights[1].T, 'output_net_hidden_layer')
```
