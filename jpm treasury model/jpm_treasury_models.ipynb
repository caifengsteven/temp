# JPMorgan US Treasury Fair-Value Models

Replication of three JPMorgan US Rates Strategy quantitative models:

1. **10-year Treasury Fair-Value Model** — regresses 10y yield on Fed policy expectations, inflation expectations, growth forecasts, Fed balance sheet, and trade policy uncertainty
2. **5s/30s Curve Fair-Value Model** — regresses the 5s30s slope on the same factors (excluding growth FRI)
3. **Butterfly Relative Value** — level-neutral and curve-neutral regression of Treasury butterflies

*Note: Set `USE_BLOOMBERG = True` if you have a Bloomberg Terminal running with xbbg installed. Otherwise the framework uses realistic synthetic data.*

```python
import warnings
warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt
import pandas as pd

# Set to True if Bloomberg Terminal + xbbg are available
USE_BLOOMBERG = False

pd.set_option("display.width", 140)
pd.set_option("display.max_columns", 20)
pd.set_option("display.float_format", lambda v: f"{v:.4f}")

print("JPMorgan Treasury Fair-Value Framework")
print("=" * 60)
```

---
## Model 1: 10-Year Treasury Fair-Value

```python
from src.model_10yr import run_10yr_model, factor_contributions, fair_value_report, plot_factor_contributions

model_10yr = run_10yr_model(use_bloomberg=USE_BLOOMBERG)

report = fair_value_report(model_10yr)
print(f"R² = {model_10yr.rsquared:.4f}  |  Obs = {report['n_observations']}  |  Date = {report['date']}")
print(f"Actual 10y Yield:  {report['actual_yield']:.4f}%")
print(f"Fair Value:        {report['fair_value_yield']:.4f}%")
print(f"Residual:          {report['residual_bp']:+.2f}bp  →  {report['signal']}")
```

```python
factor_contributions(model_10yr)
```

```python
fig = model_10yr.plot_actual_vs_fair()
plt.show()

fig = model_10yr.plot_residual()
plt.show()

fig = plot_factor_contributions(model_10yr)
plt.show()
```
